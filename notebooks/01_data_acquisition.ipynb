```python
import yfinance as yf
import pandas as pd
import numpy as np

# 1. Download raw historical S&P 500 data
print("Connecting to Yahoo Finance...")
sp500 = yf.download('^GSPC', start='2000-01-01', end='2026-06-01')

# FIX: Flatten the multi-level column index created by newer versions of yfinance
if isinstance(sp500.columns, pd.MultiIndex):
    sp500.columns = sp500.columns.droplevel(1)

# 2. Engineer the core features
# Calculate Daily Returns
sp500['Returns'] = sp500['Close'].pct_change()

# Calculate Intraday Volatility Proxy
sp500['Volatility'] = (sp500['High'] - sp500['Low']) / sp500['Open']

# Calculate Abnormal Volume (Ratio of daily volume to 20-day moving average)
sp500['Volume_20SMA'] = sp500['Volume'].rolling(window=20).mean()
sp500['Abnormal_Volume'] = sp500['Volume'] / sp500['Volume_20SMA']

# 3. Clean the dataset
clean_data = sp500[['Close', 'Returns', 'Volatility', 'Volume', 'Abnormal_Volume']].dropna()

# Display the first 5 rows to verify
print("\nData Collection and Feature Engineering Complete!")
clean_data.head()
```
