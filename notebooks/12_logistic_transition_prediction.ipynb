# Leakage-safe logistic transition prediction
Logistic regression estimates the probability of a binary future transition. Predictors are known after the close at date t; the binary target refers to t+1 through t+5. Model 1 is the baseline, Model 2 adds log abnormal volume, and Model 3 examines pre-specified volatility heterogeneity.

The 2018–2026 test labels are not used for preprocessing, model selection, or thresholds. Expanding-window validation preserves time order, and purging prevents overlapping outcome windows from crossing fold boundaries. Coefficients describe log-odds and exponentiated coefficients are odds ratios; training significance is distinct from test prediction.

ROC AUC measures ranking across both classes; PR AUC emphasizes the transition class. Brier score and log loss assess probabilistic accuracy. Precision, recall, F1, specificity, and balanced accuracy use training-selected thresholds. Calibration compares predicted with observed rates. Paired moving-block bootstrap intervals retain temporal dependence and paired model predictions.

The final evidence classification follows the pre-specified yes/no/conditional rule. Limitations include overlapping targets, estimated regimes, class imbalance, and possible structural change.
