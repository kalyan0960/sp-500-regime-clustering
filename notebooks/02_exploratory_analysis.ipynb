```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load the data you saved in the previous step
# We use '../data/' to tell the code to step out of the notebooks folder and into the data folder
df = pd.read_csv('../data/sp500_engineered_features.csv', index_col='Date', parse_dates=True)

# 2. Create a visual scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(df['Volatility'], df['Returns'], alpha=0.4, color='blue')

# 3. Add labels and formatting
plt.title('S&P 500: Daily Returns vs. Intraday Volatility (2000 - 2026)', fontsize=14)
plt.xlabel('Volatility (Intraday Price Swing)', fontsize=12)
plt.ylabel('Daily Returns', fontsize=12)
plt.axhline(0, color='black', linestyle='--', linewidth=1) # Adds a baseline at 0% returns
plt.grid(True, alpha=0.3)

plt.show()
```
